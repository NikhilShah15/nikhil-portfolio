```python
# =========================
# POWER BI EXPORT FILES
# =========================

# 1. Customers
df_customers_powerbi = df_customers[
    [
        'customer_id',
        'customer_unique_id',
        'customer_zip_code_prefix',
        'customer_city',
        'customer_state'
    ]
].copy()

df_customers_powerbi.to_csv(
    'Customers_PowerBI.csv',
    index=False
)


# 2. Orders
# Keep the original order-level data and add New/Repeated classification.
order_type_map = (
    orders_customers[
        ['order_id', 'order_type']
    ]
    .drop_duplicates('order_id')
)

df_orders_powerbi = df_orders.merge(
    order_type_map,
    on='order_id',
    how='left',
    suffixes=('', '_customer')
)

if 'order_type_customer' in df_orders_powerbi.columns:
    df_orders_powerbi['order_type'] = df_orders_powerbi['order_type_customer']
    df_orders_powerbi.drop(
        columns=['order_type_customer'],
        inplace=True
    )

df_orders_powerbi.to_csv(
    'Orders_PowerBI.csv',
    index=False
)


# 3. Order Items + Product Category
df_order_items_powerbi = (
    df_order_items
    .merge(
        df_products[
            ['product_id', 'product_category_name']
        ],
        on='product_id',
        how='left'
    )
    .merge(
        df_category_translation,
        on='product_category_name',
        how='left'
    )
)

df_order_items_powerbi.to_csv(
    'Order_Items_PowerBI.csv',
    index=False
)


# 4. Delivery + Review
df_reviews_powerbi = df_delivery_reviews[
    ['order_id', 'delivery_days', 'review_score']
].copy()

df_reviews_powerbi = df_reviews_powerbi[
    df_reviews_powerbi['delivery_days'].notna()
    & df_reviews_powerbi['review_score'].notna()
].copy()

df_reviews_powerbi['delivery_bucket'] = pd.cut(
    df_reviews_powerbi['delivery_days'],
    bins=[0, 5, 10, 15, 20, 25, 29.145],
    labels=[
        '0-5',
        '5-10',
        '10-15',
        '15-20',
        '20-25',
        '25-29.145'
    ],
    right=False
)

df_reviews_powerbi.to_csv(
    'Delivery_Reviews_PowerBI.csv',
    index=False
)


# 5. Payments
# In the original notebook, df was last loaded as the payment dataset.
df_payments_powerbi = df.copy()

df_payments_powerbi.to_csv(
    'Payments_PowerBI.csv',
    index=False
)


# 6. Payment Summary — one row per order
df_payment_summary_powerbi = (
    df_payments_powerbi
    .groupby('order_id')
    .agg(
        total_payment_value=('payment_value', 'sum'),
        payment_count=('payment_sequential', 'count'),
        payment_methods=('payment_type', 'nunique'),
        max_installments=('payment_installments', 'max')
    )
    .reset_index()
)

df_payment_summary_powerbi.to_csv(
    'Payment_Summary_PowerBI.csv',
    index=False
)


# Check the exported files
print("Power BI export files created successfully:")
print("Customers_PowerBI.csv")
print("Orders_PowerBI.csv")
print("Order_Items_PowerBI.csv")
print("Delivery_Reviews_PowerBI.csv")
print("Payments_PowerBI.csv")
print("Payment_Summary_PowerBI.csv")
```
